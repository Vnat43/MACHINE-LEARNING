# **PRISE EN MAIN DE LA BASE DE DONNEES**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use("ggplot")

df = pd.read_csv("hotels.csv", sep=",", header=0)
df
```

```python
df.describe()
```

```python
df.info()
```

## Drop of the 'company' column 

```python
df.drop("company", axis=1, inplace=True)
df
```

## Drop of the lines with missing values + negative value of ADR

```python
df.adr.replace(-6.38, np.nan, inplace=True)
df_drop = df.dropna()
print(df_drop.shape)

df_drop.info()
```

## Drop of the 'country' and 'reservation_status_date' columns because not convertible to int

```python
print(df_drop.groupby("hotel").size())
print(df_drop.groupby("is_canceled").size())
print(df_drop.groupby("arrival_date_month").size())
print(df_drop.groupby("meal").size())
print(df_drop.groupby("country").size())
print(df_drop.groupby("market_segment").size())
print(df_drop.groupby("distribution_channel").size())
print(df_drop.groupby("reserved_room_type").size())
print(df_drop.groupby("assigned_room_type").size())
print(df_drop.groupby("deposit_type").size())
print(df_drop.groupby("agent").size())
print(df_drop.groupby("customer_type").size())
print(df_drop.groupby("adr").size())
print(df_drop.groupby("reservation_status").size())
print(df_drop.groupby("reservation_status_date").size())
```

```python
df_drop.drop("country", axis=1, inplace=True)
df_drop.drop("reservation_status", axis=1, inplace=True)
df_drop.drop("reservation_status_date", axis=1, inplace=True)
```

## ENCODING OF THE CHARACTER COLUMNS

```python
df_char = df_drop.copy()
df_char["hotel"] = df_char["hotel"].map({"City Hotel":1, "Resort Hotel":2})
df_char["arrival_date_month"] = df_char["arrival_date_month"].map({"January":1, "February":2, "March":3, "April":4, "May":5, "June":6, "July":7, "August":8, "September":9, "October":10, "November":11, "December":12})
df_char["meal"] = df_char["meal"].map({"BB":1, "FB":2, "HB":3, "SC":4, "Undefined":5})  #On prend SC=Undefined
df_char["distribution_channel"] = df_char["distribution_channel"].map({"Corporate":1, "Direct":2, "GDS":3, "TA/TO":4})
df_char["market_segment"] = df_char["market_segment"].map({"Aviation":1, "Complementary":2, "Corporate":3, "Direct":4, "Groups":5, "Offline TA/TO":6, "Online TA":7})
df_char["reserved_room_type"] = df_char["reserved_room_type"].map({"A":1, "B":2, "C":3, "D":4, "E":5, "F":6, "G":7, "H":8})
df_char["assigned_room_type"] = df_char["assigned_room_type"].map({"A":1, "B":2, "C":3, "D":4, "E":5, "F":6, "G":7, "H":8, "I":1, "K":2})
df_char["deposit_type"] = df_char["deposit_type"].map({"No Deposit":1, "Non Refund":2, "Refundable":3})
df_char["customer_type"] = df_char["customer_type"].map({"Contract":1, "Group":2, "Transient":3, "Transient-Party":4})
```

# **MODEL WITH ALL THE VALUES**

```python
X = df_char.drop("is_canceled", axis=1)
Y = df_char.is_canceled
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=21, stratify=Y)
```

## KNEIGHBORS

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=8)
knn.fit(X_train, Y_train)
```

```python
Y_pred_test = knn.predict(X_test)
Y_pred_train = knn.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score as accuracy

print("Accuracy on test set is: ", round(accuracy(Y_test, Y_pred_test) ,5))
print("Accuracy on train set is: ", round(accuracy(Y_train, Y_pred_train) ,5))

from sklearn.metrics import confusion_matrix

print(confusion_matrix(Y_test, Y_pred_test))
```

# **MODEL WITH HYPERPARAMETER TUNING**

## KNeighbors

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline

pipe = Pipeline([("knn", knn_drop)])
parameters = [{"knn__n_neighbors":np.arange(1,50)}]   #Test des hyperparamètres

crossval = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
grid_knn = GridSearchCV(pipe, param_grid=parameters, cv=crossval, n_jobs=-1)

grid_knn.fit(X_train_drop, Y_train)
```

```python
print(grid_knn.best_params_)
print(grid_knn.best_score_)

from sklearn.metrics import accuracy_score as accuracy

Y_pred_test_gridcv = grid_knn.best_estimator_.predict(X_test_drop)
Y_pred_train_gridcv = grid_knn.best_estimator_.predict(X_train_drop)
print(accuracy(Y_test, Y_pred_test_gridcv))
print(accuracy(Y_train, Y_pred_train_gridcv))
```

```python
scores = pd.DataFrame(grid_knn.cv_results_)
scores = scores.sort_values(by="rank_test_score").set_index("rank_test_score")
scores.head()
```

## SVC

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline

pipe = Pipeline([("scaler", MinMaxScaler()), ("svc", svc_drop)])

parameters = [{"svc__kernel": ["linear"], "svc__C": [0.1, 1, 10]},
              {"svc__kernel": ["rbf"], "svc__C": [0.1, 1, 10], "svc__gamma":[0.1, 1, 10]},
              {"svc__kernel": ["poly"], "svc__C": [0.1, 1, 10], "svc__gamma":[0.1, 1, 10], "svc__degree": np.arange(1,5,1)}]

crossval = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

grid_svc = GridSearchCV(pipe, param_grid = parameters, cv=crossval, n_jobs=-1)
grid_svc.fit(X_train_drop, Y_train)
```

```python
Y_pred_train_cv_svc = grid_svc.best_estimator_.predict(X_train_drop)
Y_pred_test_cv_svc = grid_svc.best_estimator_.predict(X_test_drop)

print(grid_svc.best_params_)
print(grid_svc.best_score_)

print(accuracy(Y_train, Y_pred_train_cv_svc))
print(accuracy(Y_test, Y_pred_test_cv_svc))

scores = pd.DataFrame(grid_svc.cv_results_)
scores = scores.sort_values(by="rank_test_score").set_index("rank_test_score")
scores.head()
```
